```python
LAST_NAME = "Jorvina"
STUDENT_ID = "TUPM-26-1946"
FAVORITE_ARTIST = "Miguel"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(char.upper()) - ord("A") + 1 for char in FAVORITE_ARTIST if char.isalpha())

print("Seed Digit:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
print("Artist Length:", ARTIST_LENGTH)
print("Artist Sum:", ARTIST_SUM)
```

```python
def linear_operations(n):
    count = 0
    for _ in range(n):
        count += 1
    return count

for n in [10, 100, 1000, 5000, 10000]:
    print(f"n={n:5d} | operations={linear_operations(n):5d}")
```

```python
def log_ops(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count

for n in [10, 100, 1000]:
    print(f"n={n:4d} | O(1)=1 | O(log n)={log_ops(n):2d} | O(n)={n:4d} | O(n^2)={n*n:7d}")
```

```python
def constant_space(n):
    total = 0
    for value in range(n):
        total += value
    return total

def linear_space(n):
    return list(range(n))

for n in [10, 100, 1000, 5000]:
    data = linear_space(n)
    print(f"n={n:5d} | O(1) extra space: constant | O(n) stored items: {len(data)}")
```

```python
def search_case(data, target):
    comparisons = 0
    for value in data:
        comparisons += 1
        if value == target:
            return comparisons
    return comparisons

data = [11, 22, 33, 44, 55]
for case, target in {"Best": 11, "Average": 33, "Worst": 99}.items():
    print(f"{case} Case: {search_case(data, target)} comparisons")
```

```python
def linear_search(data, target):
    comparisons = 0
    for index, value in enumerate(data):
        comparisons += 1
        if value == target:
            return index, comparisons
    return -1, comparisons

data = list(range(1, 21))
target = SEED_DIGIT + 1
index, comparisons = linear_search(data, target)
print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
def binary_search(data, target):
    low, high = 0, len(data) - 1
    comparisons = 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, comparisons

index, comparisons = binary_search(data, target)
print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
search_data = list(range(1, 101))
search_target = ((ID_SUM + SEED_DIGIT) % 100) + 1
linear_index, linear_comps = linear_search(search_data, search_target)
binary_index, binary_comps = binary_search(search_data, search_target)
print("Target:", search_target)
print(f"Linear -> index={linear_index}, comparisons={linear_comps}")
print(f"Binary -> index={binary_index}, comparisons={binary_comps}")
print("Comparison Reduction:", linear_comps - binary_comps)
```

```python
def bubble_sort(data):
    a = data.copy()
    comparisons = swaps = 0
    for i in range(len(a) - 1):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

sort_data = [(ID_SUM * 3) % 100, (SEED_DIGIT * 11) % 100,
             (NAME_LENGTH * 7) % 100, 42, 17, 89, 31, 56]
bubble_result = bubble_sort(sort_data)
print("Original:", sort_data)
print("Sorted:", bubble_result[0])
print("Comparisons:", bubble_result[1])
print("Swaps:", bubble_result[2])
```

```python
def selection_sort(data):
    a = data.copy()
    comparisons = swaps = 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

selection_result = selection_sort(sort_data)
print("Sorted:", selection_result[0])
print("Comparisons:", selection_result[1])
print("Swaps:", selection_result[2])
```

```python
def insertion_sort(data):
    a = data.copy()
    comparisons = shifts = 0
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

insertion_result = insertion_sort(sort_data)
print("Sorted:", insertion_result[0])
print("Comparisons:", insertion_result[1])
print("Shifts:", insertion_result[2])
```

```python
import time

def measure(sort_function, data, trials=5):
    times = []
    result = None
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / len(times)

data_size = 100 + ID_SUM + 19
performance_data = [((i * 34 + SEED_DIGIT * 7) % 1000) for i in range(data_size)]

bubble_result, bubble_time = measure(bubble_sort, performance_data)
selection_result, selection_time = measure(selection_sort, performance_data)
insertion_result, insertion_time = measure(insertion_sort, performance_data)

print("Dataset Size:", data_size)
print(f"Bubble Sort: {bubble_time:.8f} s")
print(f"Selection Sort: {selection_time:.8f} s")
print(f"Insertion Sort: {insertion_time:.8f} s")
```

```python
results = [
    ("Bubble Sort", bubble_result, bubble_time),
    ("Selection Sort", selection_result, selection_time),
    ("Insertion Sort", insertion_result, insertion_time),
]
print(f"{'Algorithm':<18}{'Comparisons':>12}{'Movements':>12}{'Time (s)':>14}")
for name, result, elapsed in results:
    print(f"{name:<18}{result[1]:>12}{result[2]:>12}{elapsed:>14.8f}")
```
