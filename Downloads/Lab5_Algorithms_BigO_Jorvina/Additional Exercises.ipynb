```python
LAST_NAME = "Jorvina"
STUDENT_ID = "TUPM-26-1946"
FAVORITE_ARTIST = "Miguel"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(
    ord(char.upper()) - ord("A") + 1
    for char in FAVORITE_ARTIST
    if char.isalpha()
)

print("SEED_DIGIT:", SEED_DIGIT)
print("ID_SUM:", ID_SUM)
print("NAME_LENGTH:", NAME_LENGTH)
print("ARTIST_LENGTH:", ARTIST_LENGTH)
print("ARTIST_SUM:", ARTIST_SUM)
```

```python
def linear_search(data, target):
    comparisons = 0

    for index, value in enumerate(data):
        comparisons += 1

        if value == target:
            return index, comparisons

    return -1, comparisons
```

```python
def bubble_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(len(arr)):
        for j in range(0, len(arr) - i - 1):
            comparisons += 1

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                movements += 3

    return arr, comparisons, movements


def selection_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(len(arr)):
        min_index = i

        for j in range(i + 1, len(arr)):
            comparisons += 1

            if arr[j] < arr[min_index]:
                min_index = j

        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            movements += 3

    return arr, comparisons, movements


def insertion_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if arr[j] > key:
                arr[j + 1] = arr[j]
                movements += 1
                j -= 1
            else:
                break

        arr[j + 1] = key
        movements += 1

    return arr, comparisons, movements
```

```python
import time

def measure(function, values, trials=5):
    total_time = 0.0
    result = None

    for _ in range(trials):
        start = time.perf_counter()
        result = function(values)
        end = time.perf_counter()

        total_time += end - start

    average_time = total_time / trials
    return result, average_time
```

```python
orders = [
    ((i * 17 + ID_SUM * 3 + SEED_DIGIT) % 100)
    for i in range(12)
]

benchmark_sorting_conditions(orders)
```

```python
def rotate_list(values, rotation_point):
    if not values:
        return []
    k = rotation_point % len(values)
    return values[k:] + values[:k]

def rotated_binary_search(data, target):
    low, high = 0, len(data) - 1
    comparisons = 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        # Determine which half is sorted, then keep the half that can contain target.
        if data[low] <= data[mid]:
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons

original_incidents = list(range(1000, 1020))
rotation_point = SEED_DIGIT % len(original_incidents)
rotated_incidents = rotate_list(original_incidents, rotation_point)
target_incident = 1000 + ((ID_SUM + ARTIST_SUM) % len(original_incidents))

rot_index, rot_comparisons = rotated_binary_search(rotated_incidents, target_incident)
lin_index, lin_comparisons = linear_search(rotated_incidents, target_incident)

print("Original Dataset:", original_incidents)
print("Rotation Point:", rotation_point)
print("Rotated Dataset:", rotated_incidents)
print("Target Incident:", target_incident)
print("Rotated Binary Search -> index:", rot_index, "| comparisons:", rot_comparisons)
print("Linear Search -> index:", lin_index, "| comparisons:", lin_comparisons)
print("Target Found:", rot_index != -1)
print("Time Complexity: O(log n) average for rotated binary search; O(n) for linear search")
print("Space Complexity: O(1) auxiliary space for both searches")
```

```python
def merge_sorted(a, b):
    i = j = comparisons = 0
    merged = []
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1
    merged.extend(a[i:])
    merged.extend(b[j:])
    return merged, comparisons

department_a = sorted({SEED_DIGIT * 10 + i * 3 for i in range(10)})
department_b = sorted({ARTIST_LENGTH * 10 + NAME_LENGTH + i * 4 for i in range(10)})
consolidated, merge_comparisons = merge_sorted(department_a, department_b)

print("Department A:", department_a)
print("Department B:", department_b)
print("Consolidated Dataset:", consolidated)
print("Records before consolidation:", len(department_a) + len(department_b))
print("Records after consolidation:", len(consolidated))
print("Comparison Count:", merge_comparisons)
print("Ordering Verified:", consolidated == sorted(department_a + department_b))
print("Time Complexity: O(n + m)")
print("Space Complexity: O(n + m) for the consolidated list")
```

```python
import random

def benchmark_sorting_conditions(base_orders, trials=3):
    conditions = {
        "Already sorted": sorted(base_orders),
        "Reverse ordered": sorted(base_orders, reverse=True),
        "Random order": random.Random(SEED_DIGIT).sample(base_orders, len(base_orders)),
    }
    algorithms = [
        ("Bubble Sort", bubble_sort),
        ("Selection Sort", selection_sort),
        ("Insertion Sort", insertion_sort),
    ]
    print("Generated order dataset:", base_orders)
    for condition, values in conditions.items():
        print(f"\n{condition}:")
        print("Input:", values)
        for name, function in algorithms:
            result, elapsed = measure(function, values, trials=trials)
            is_sorted = result[0] == sorted(values)
            print(f"{name:<16} comparisons={result[1]:4d} movements={result[2]:4d} "
                  f"average_time={elapsed:.8f}s sorted={is_sorted}")

orders = [((i * 17 + ID_SUM * 3 + SEED_DIGIT) % 100) for i in range(12)]
benchmark_sorting_conditions(orders)
```
